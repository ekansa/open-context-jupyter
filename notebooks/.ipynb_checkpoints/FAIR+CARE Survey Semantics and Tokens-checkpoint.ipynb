# Some Semantic Grouping Analysis of the FAIR-CARE Survey Text Responses

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

import os

import pandas as pd

from slugify import slugify

# Get the root_path for this jupyter notebook repo.
repo_path = os.path.dirname(os.path.abspath(os.getcwd()))
# Use this path to save the wordcloud outputs
wordcloud_1_path = os.path.join(
    repo_path, 'files', 'IMLS-FAIR-CARE-Survey', 'wordcloud-figs-1',
)

processed_survey_path = '/home/ekansa/oc-data/fair-care-survey-processed.csv' # Keep this OUT of version control, has sensitive info
token_freq_path = '/home/ekansa/oc-data/fair-care-token-freq.csv'
ngram_freq_path = '/home/ekansa/oc-data/fair-care-ngram-freq.csv'
semantic_clusters_path = '/home/ekansa/oc-data/fair-care-semantic-clusters.csv'


df = pd.read_csv(processed_survey_path, low_memory=False)
df_token = pd.read_csv(token_freq_path, low_memory=False)
df_ngram = pd.read_csv(ngram_freq_path, low_memory=False)
df_semantic = pd.read_csv(semantic_clusters_path, low_memory=False)

print(f'FAIR+CARE survey has {len(df.index)} rows')
print(f'FAIR+CARE survey token extract has {len(df_token.index)} rows')
print(f'FAIR+CARE survey ngram extract has {len(df_ngram.index)} rows')
print(f'FAIR+CARE survey semantic groupings has {len(df_semantic.index)} rows')
```

```python
gen_group_cols = ['orig_column_index', 'column', 'semantic_group_st',]
use_index = df_semantic['orig_column_index'] >= 38
df_semantic_gp1 = df_semantic[use_index ][gen_group_cols].groupby(gen_group_cols, as_index=False).first()
df_semantic_gp1.sort_values(by=['orig_column_index', 'semantic_group_st',], ascending=[True, True], inplace=True)


# Repeatly used ngrams
print(f'There are {len(df_semantic_gp1.index)} unique semantic groups across all text fields')

df_semantic_gp1.head(100)
```

```python
# Now Make WordClouds for each column, and each cluster
def create_wordcloud(text_dict, title=None, save_dir=wordcloud_1_path):
    wc = WordCloud(background_color="white", max_words=1000)
    wc.generate_from_frequencies(text_dict)
    plt.imshow(wc, interpolation="bilinear")
    plt.axis("off")
    if title:
        plt.title(title)
    slug_file = slugify(title)
    filename = f'{slug_file}.png'
    f_path = os.path.join(save_dir, filename)
    plt.savefig(f_path, bbox_inches='tight')
    plt.close()
    print(f'Saved figure: {filename}')

token_join_cols = ['Response ID', 'token', 'token_count', 'orig_column_index',]
semantic_join_cols = [
    'Response ID', 
    'orig_column_index', 
    'Question Number', 
    'Section', 
    'Sub-Section', 
    'column', 
    'response', 
    'semantic_group_st',
    'semantic_agglom_cluster_id',
    'bert_name_plain', 
    'bert_top_words_plain',
]
df_wc = pd.merge(df_token[token_join_cols], df_semantic[semantic_join_cols], on=['Response ID', 'orig_column_index',])

```

```python
use_index = df_wc['orig_column_index'] >= 38
df_wc_grp = df_wc[use_index].groupby(
    ['orig_column_index', 'column', 'semantic_group_st', 'token',], 
    as_index=False
).agg(
    {
        'Question Number': 'first',
        'Section': 'first',
        'Sub-Section': 'first',
        'token_count': 'sum',
    }
)

for col in df_wc_grp['column'].unique().tolist():
    col_index = df_wc_grp['column'] == col
    sm_groups = df_wc_grp[col_index]['semantic_group_st'].unique().tolist()
    for sm_group in sm_groups:
        col_sm_index = col_index & ( df_wc_grp['semantic_group_st'] == sm_group)
        total_kw_count = df_wc_grp[col_sm_index]['token_count'].sum()
        question = df_wc_grp[col_sm_index]['Question Number'].iloc[0]
        question = int(question)
        section = df_wc_grp[col_sm_index]['Section'].iloc[0]
        sub_section = df_wc_grp[col_sm_index]['Sub-Section'].iloc[0]
        title = f'{section} -- {sub_section} ({question}) {col} [Group {sm_group}]'
        text_dict = {}
        for _, row in df_wc_grp[col_sm_index].iterrows():
            if row['token'] == 'data':
                # not very interesting
                continue
            text_dict[row['token']] = row['token_count']
        create_wordcloud(text_dict, title)

```
