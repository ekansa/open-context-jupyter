```python
import os

import numpy as np
import pandas as pd

import re

import squarify

from slugify import slugify

# Get the root_path for this jupyter notebook repo.
repo_path = os.path.dirname(os.path.abspath(os.getcwd()))
col_config_path = os.path.join(
    repo_path, 'files', 'IMLS-FAIR-CARE-Survey', 'imls-fair-care-survey-columns-config.csv',
)

r_files_path = '/home/ekansa/oc-data/fair_care_in_r'

file_paths = []
for dirpath, _, filenames in os.walk(r_files_path):
    for filename in filenames:
        if not filename.endswith('.csv'):
            # Skip not csv file.
            continue
        file_path = os.path.join(dirpath, filename)
        file_paths.append(file_path)
        
```

```python
def make_column_r_safe(column):
    """Make a column name safe for R"""
    slug_col = slugify(column.strip())
    return slug_col.replace('-', '_')


def make_columns_r_safe(file_path):
    df = pd.read_csv(file_path)
    renames = {col:  make_column_r_safe(col) for col in df.columns.tolist()}
    df.rename(columns=renames, inplace=True)
    df.to_csv(file_path, index=False)
    print(f'Make {(file_path)} save for R')


for file_path in file_paths:
    make_columns_r_safe(file_path)
```

```python
# Now add a r_column to the fair-care-survey-col-summary.csv
path =  os.path.join(r_files_path, 'fair-care-survey-col-summary.csv')
df = pd.read_csv(path)
last_cols = df.columns.tolist()
df['column_r_safe'] = df['column'].apply(make_column_r_safe)
cols = ['column_r_safe'] + last_cols
df[cols].to_csv(path, index=False)
```
