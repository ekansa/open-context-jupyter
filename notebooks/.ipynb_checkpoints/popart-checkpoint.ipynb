```python
import os

import numpy as np
import pandas as pd

path =  '/home/ekansa/oc-data/GCI-Reference-collection-PopArt.csv'
col_meta_path = '/home/ekansa/oc-data/GCI-Reference-collection-PopArt-column-metadata.csv'
df = pd.read_csv(path)

tsr_path = '/home/ekansa/oc-data/TSR.csv'
tsr_col_meta_path = '/home/ekansa/oc-data/TSR-column-metadata.csv'
df_tsr = pd.read_csv(tsr_path)
```

```python
col_types = [
    ['Analysis Lab',],
    ['Analytical Result',],
    ['Date',],
    ['File',],
    ['Instrument',],
    ['Instrument parameters', 'instrument param'],
    ['Operator', 'Chromatografist', 'Spectrometrist', ], 
    ['Sample Preparation',],
    ['Shift',],
    ['absorption bands',],
    ['Spectral Manipulation',],
]

col_method = [
    'DS',
    'EA',
    'IR',

    'DSC',
    'EGA',
    'NIR',
    'TGA',

    'FTIR',

    'PY MS',
    'RAMAN',
    'ChemiL',
    
    'Py GCMS',
    'TDS THz',

    'PROFILOMETRY',
]

col_institutions = [

    'VAM',
    'MORANA',
    'ATR MORANA',
    'C2RMF',
    'GCI',
    'RCE',
    'UCL',
    'PISAS',
]

cols = df.columns.tolist()
rows = []
for col in cols:
    row = df[col].describe()
    method = ''
    for act_meth in col_method:
        if f'{act_meth} ' in col:
            method = act_meth
    row['analysis_method'] = method
    col_type = ''
    l_col = col.lower()
    for col_type_list in col_types:
        main_col_type = col_type_list[0].lower()
        for act_col_type in col_type_list:
            act_col_type = act_col_type.lower()
            if act_col_type in l_col:
                col_type = main_col_type
    row['analysis_metadata_type'] = col_type
    inst = ''
    for act_inst in col_institutions:
        if col.endswith(act_inst):
            inst = act_inst
    row['analysis_institution'] = inst
    rows.append(row)

df_cols = pd.DataFrame(rows)
col_cols = df_cols.columns.tolist()
df_cols['column'] = df_cols.index
df_cols = df_cols.reset_index()
df_cols = df_cols[(['column'] + col_cols)]
```

```python
df_cols.to_csv(col_meta_path, index=False)
df_cols.head(100)
```

```python
tsr_rows = []
for col in df_tsr.columns.tolist():
    row = df_tsr[col].describe()
    tsr_rows.append(row)

df_tsr_cols = pd.DataFrame(tsr_rows)
tsr_col_cols = df_tsr_cols.columns.tolist()
df_tsr_cols['column'] = df_tsr_cols.index
df_tsr_cols = df_tsr_cols.reset_index()
df_tsr_cols = df_tsr_cols[(['column'] + tsr_col_cols)]

df_tsr_cols.to_csv(tsr_col_meta_path, index=False)
df_tsr_cols.head(100)
```
