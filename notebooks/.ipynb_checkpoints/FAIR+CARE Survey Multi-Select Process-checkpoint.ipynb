```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

import os

import numpy as np
import pandas as pd

import re

# Get the root_path for this jupyter notebook repo.
repo_path = os.path.dirname(os.path.abspath(os.getcwd()))
col_config_path = os.path.join(
    repo_path, 'files', 'IMLS-FAIR-CARE-Survey', 'imls-fair-care-survey-columns-config.csv',
)

processed_survey_path = '/home/ekansa/oc-data/fair-care-survey-processed.csv' # Keep this OUT of version control, has sensitive info
wide_survey_path = '/home/ekansa/oc-data/fair-care-survey-multi-select-expanded.csv' # Keep this OUT of version control, has sensitive info
col_summary_path = '/home/ekansa/oc-data/fair-care-survey-col-summary.csv'

df_config = pd.read_csv(col_config_path, low_memory=False)
df_orig = pd.read_csv(processed_survey_path, low_memory=False)
df_col_sum = pd.read_csv(col_summary_path, low_memory=False)

print(f'FAIR+CARE survey has {len(df_orig.index)} rows')
print(f'FAIR+CARE survey column summary has {len(df_col_sum.index)} rows')
```

```python
# Select the columns with select option (not free-text)
cols_with_free_text = (
    ((df_col_sum['count_unique'] / df_col_sum['count_not_blank']) >= 0.33) 
    & (df_col_sum['max_word_count'] > (df_col_sum['mean_word_count'] * 1.5) )
)
cols_with_select_opts = (
    (df_col_sum['orig_column_index'] >= 18)
    & ((df_col_sum['count_unique'] / df_col_sum['count_not_blank']) < 0.33)
    & ~(df_col_sum['column'].str.endswith('Text'))
    & ~(cols_with_free_text)
)
select_cols = df_col_sum[cols_with_select_opts]['column'].unique().tolist()
```

```python
def reorder_new_columns_after_col_in_df(col, new_columns, df):
    """Reorders new columns in a dataframe to follow a specific column"""
    if not col in df.columns.tolist():
        raise ValueError(f'{col} is not in the dataframe df!')
    before_cols = []
    after_cols = []
    col_found = False
    for old_col in df.columns.tolist():
        if not col_found:
            before_cols.append(old_col)
        else:
            after_cols.append(old_col)
        if old_col == col:
            col_found = True
    all_cols = before_cols + new_columns + after_cols
    reorder_cols = [c for c in all_cols if c in df.columns.tolist()]
    # Return the dataframe with all the columns reordered in the proper way
    return df[reorder_cols]


def remove_spurious_delim(text):
    """Removes spurious delimerters from within parentheses"""
    # This regex finds commas between parentheses
    return re.sub(r'\(([^)]*?),([^)]*?)\)', lambda x: x.group(0).replace(',', ''), text)
        


def make_str_val_column_safe(str_val):
    """Makes a string value safe for use in a column name"""
    str_val = str_val.strip()
    # Comment this out. Let's keep punctuation, just no quotes
    # str_val = re.sub(r'[^\w\s]', '', str_val)
    bad_chars = ['"', "'",]
    for bad_char in bad_chars:
        str_val.replace(bad_char, '')
    return str_val


def split_val_in_col_val(check_split_val, col_val):
    """Checks to see if a split_val is in a multi-selection column value"""
    if check_split_val == col_val:
        return True
    if not isinstance(col_val, str):
        return False
    check_split_val = str(check_split_val)
    col_val = remove_spurious_delim(col_val)
    for col_val_part in col_val.split(','):
        col_val_part = make_str_val_column_safe(col_val_part)
        if col_val_part == check_split_val:
            return True
    return False
    
    
```

```python
# Add new columns for multiple select options
new_dfs = []
for sel_col in select_cols:
    if not sel_col in df_orig.columns.tolist():
        # for some reason, we don't have this column in our dataset
        continue
    df = df_orig[['Response ID', sel_col,]].copy()
    col_index = ~df[sel_col].isnull()
    col_unique_vals = df[col_index][sel_col].unique().tolist()
    print(f'{sel_col} has {len(col_unique_vals)} unique values')
    col_unique_parts = set()
    for col_val in col_unique_vals:
        if not isinstance(col_val, str):
            col_unique_parts.add(str(col_val))
            continue
        col_val = remove_spurious_delim(col_val)
        for col_val_part in col_val.split(','):
            split_val = make_str_val_column_safe(col_val_part)
            col_unique_parts.add(split_val)
    if len(col_unique_vals) == len(col_unique_parts):
        # This column has likely has a single select option or nobody did multiple select
        # So we don't have to split this into multiple columns
        new_dfs.append(df)
        continue
    if len(col_unique_parts) < 2:
        # There's no need to split this column, there aren't enough unique parts.
        new_dfs.append(df)
        continue
    print(f'{sel_col} has {len(col_unique_vals)} unique values, with {len(col_unique_parts)} select options')
    print(f'Make new columns for {col_unique_parts}')
    new_columns = []
    for split_val in col_unique_parts:
        split_val = str(split_val)
        # Make a new column for this split_value, and add it to the dataframe with null data.
        new_col = f'{sel_col}::{split_val}::TF'
        df[new_col] = np.nan
        df[new_col] = df[new_col].astype('str')
        # Make the column False (default) for non-blank responses to this column
        df.loc[col_index, new_col] = 'False'
        df.loc[~col_index, new_col] = np.nan 
    for split_val in col_unique_parts:
        split_val = str(split_val)
        # Make a new column for this split_value, and add it to the dataframe with null data.
        new_col = f'{sel_col}::{split_val}::TF'
        # Now populate the data for this split value.
        for col_val in col_unique_vals:
            col_val_index = (df[sel_col] == col_val) & col_index
            if not split_val_in_col_val(check_split_val=split_val, col_val=col_val):
                # The split_val is not part of this col_val, so don't update
                # the dataframe new_col with True values.
                continue
            # The current split_val is in the multi-select values of col_val,
            # make the new_col be True for these cases.
            df.loc[col_val_index, new_col] = 'True'
        new_columns.append(new_col)
    new_dfs.append(df)
    print(f'Our list of new_dfs has {len(new_dfs)} members')
    print('\n')

# Merge all of our new dataframes together
df_all = new_dfs[0]
for df in new_dfs[1:]:
    df_all = pd.merge(df_all, df, on=['Response ID'])

df_all.to_csv(wide_survey_path, index=False)
print(f'The new wide dataframe has {len(df_all.columns.tolist())} columns')
```
